# Symbolic simplicial tensor examples

The examples use the shared SageMath-compatible package implementation. SageMath is used when available, and SymPy otherwise. The CoCalc adapter retains the `tensor_data` constructor keyword and matrix display methods.


```python
import numpy as np
from simplicial_tensors.notebooks.cocalc_compatible_tensor_ops import (
    SymbolicTensor,
    check_symbolic_corrections,
    correction_rank,
    n_hypergroupoid_conjecture,
    test_symbolic_n_hypergroupoid,
)
```

```python
tensor = SymbolicTensor((3, 3))
horn = tensor.horn(1)
selected = tensor.filler(horn, 1)
assert correction_rank(tensor, selected) == 2
assert not tensor.n_hypergroupoid_comparison(outer_horns=True)
print(tensor.to_latex())
print("Correction rank:", correction_rank(tensor, selected))
```

```python
zero = SymbolicTensor((2, 2), init_type="zeros")
assert zero.is_degen()
assert not zero.n_hypergroupoid_comparison(outer_horns=True, allow_degen=True)
edge = SymbolicTensor((2, 2), tensor_data=[[0, 1], [0, 0]])
degenerate_sum = edge.degen(0) + edge.degen(1)
normalized, terms = degenerate_sum.decompose_degen()
assert degenerate_sum.is_degen()
assert np.all(normalized.tensor == 0)
reconstructed = normalized
for lower, index in terms:
    reconstructed = reconstructed + lower.degen(index)
assert np.array_equal(reconstructed.tensor, degenerate_sum.tensor)
```

```python
def compare_shapes(shapes):
    """Compare selected fillers with the horn-kernel criterion for supplied shapes."""
    return [test_symbolic_n_hypergroupoid(tuple(shape), verbose=False)[:2]
            for shape in shapes]

assert compare_shapes([(3, 3)]) == [(False, False)]
# Additional shapes, such as (4, 5, 6), can be supplied explicitly.
```
